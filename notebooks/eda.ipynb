```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load Cleaned Data
df = pd.read_csv(r'C:\Users\MANGESH\customer churn prediction\data\cleaned_telco.csv')
```

```python
#Basic Statistics
df.describe()
```

```python
#Churn Distribution
sns.countplot(x='Churn', data=df)
plt.title("Churn Distribution")
plt.show()
```

```python
#Churn Percentage
df['Churn'].value_counts(normalize=True)*100
```

```python
#Gender vs Churn
sns.countplot(x='gender', hue='Churn', data=df)
plt.show()
```

```python
#Contract vs Churn
sns.countplot(x='Contract', hue='Churn', data=df)
plt.xticks(rotation=20)
plt.show()
```

```python
#Internet Service vs Churn
sns.countplot(x='InternetService', hue='Churn', data=df)
plt.show()
```

```python
#Monthly Charges Distribution
sns.histplot(df['MonthlyCharges'], bins=30)
plt.show()
```

```python
#Tenure Distribution
sns.histplot(df['tenure'], bins=30)
plt.show()
```

```python
#Numerical Correlation
num_cols = ['SeniorCitizen',
            'tenure',
            'MonthlyCharges',
            'TotalCharges',
            'Churn']

corr = df[num_cols].corr()

plt.figure(figsize=(8,5))
sns.heatmap(corr,
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("""
1. Month-to-month customers churn more.
2. Customers with shorter tenure churn more.
3. Higher monthly charges are associated with churn.
4. Long-term contracts reduce churn.
""")
```
