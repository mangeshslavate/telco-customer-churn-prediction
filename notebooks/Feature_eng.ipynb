```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\MANGESH\customer churn prediction\data\cleaned_telco.csv")

df.head()
```

```python
df.info()
```

```python
df["AvgMonthlySpend"] = (
    df["TotalCharges"] /
    (df["tenure"] + 1)
)
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0,12,24,48,72],
    labels=[
        "0-12",
        "13-24",
        "25-48",
        "49-72"
    ]
)
```

```python
df["ChargeCategory"] = pd.cut(
    df["MonthlyCharges"],
    bins=3,
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
df.head()
```

```python
df.to_csv(
    "../data/feature_engineered_telco.csv",
    index=False
)

print("Feature Engineering Completed")
```
